```python
import re


password=input("Enter password:")


if len(password)<8:
    print("Weak Password")

elif re.search("[A-Z]",password) and re.search("[0-9]",password):
    print("Strong Password")

else:
    print("Medium Password")
```

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB


messages=[
"Win money now",
"Meeting at 5 pm",
"Free prize",
"Project discussion"
]


labels=[
"spam",
"normal",
"spam",
"normal"
]


vector=CountVectorizer()

x=vector.fit_transform(messages)


model=MultinomialNB()

model.fit(x,labels)


test=input("Enter message:")

result=model.predict(
vector.transform([test])
)


print(result[0])
```
