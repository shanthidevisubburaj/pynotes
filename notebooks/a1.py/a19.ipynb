```python
class BankAccount:
    def __init__(self, name, balance):
        self.name = name
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print(f"Deposited ₹{amount}")

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print(f"Withdrawn ₹{amount}")
        else:
            print("Insufficient Balance")

    def display(self):
        print(f"\nAccount Holder: {self.name}")
        print(f"Current Balance: ₹{self.balance}")

acc = BankAccount("Shanthi", 10000)

acc.deposit(5000)
acc.withdraw(2000)
acc.display()
```

```python
books = {
    "Python": True,
    "Java": True,
    "AI": True
}

while True:
    print("\n1.View Books")
    print("2.Borrow Book")
    print("3.Return Book")
    print("4.Exit")

    choice = input("Choice: ")

    if choice == "1":
        for book, status in books.items():
            print(book, "-", "Available" if status else "Borrowed")

    elif choice == "2":
        book = input("Book Name: ")
        if books.get(book):
            books[book] = False
            print("Book Borrowed")
        else:
            print("Book Not Available")

    elif choice == "3":
        book = input("Book Name: ")
        books[book] = True
        print("Book Returned")

    else:
        break
```

```python
passwords = {}

while True:
    print("\n1.Add")
    print("2.View")
    print("3.Exit")

    ch = input("Choice: ")

    if ch == "1":
        website = input("Website: ")
        pwd = input("Password: ")
        passwords[website] = pwd

    elif ch == "2":
        for site, pwd in passwords.items():
            print(site, ":", pwd)

    else:
        break
```
