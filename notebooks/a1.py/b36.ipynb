```python
text=input("Enter message: ")

key=3

encrypted=""

for ch in text:
    encrypted += chr(ord(ch)+key)

print("Encrypted:",encrypted)


decrypted=""

for ch in encrypted:
    decrypted += chr(ord(ch)-key)

print("Decrypted:",decrypted)
```

```python
import pandas as pd

data={
    "City":["Chennai","Salem","Madurai"],
    "Temperature":[35,33,36]
}

df=pd.DataFrame(data)

print(df)

print("\nAverage Temperature:",
      df["Temperature"].mean())

print("\nHighest Temperature:")
print(df.loc[df["Temperature"].idxmax()])
```

```python
responses={
"hello":"Hi! How can I help?",
"name":"I am Python AI Bot",
"bye":"Goodbye"
}


while True:

    user=input("You: ").lower()


    if user in responses:
        print("Bot:",responses[user])

    else:
        print("Bot: I don't understand")

    if user=="bye":
        break
```
