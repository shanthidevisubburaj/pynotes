```python
balance = 5000

while True:
    print("\n1.Check Balance")
    print("2.Deposit")
    print("3.Withdraw")
    print("4.Exit")

    ch = int(input())

    if ch == 1:
        print("Balance:", balance)

    elif ch == 2:
        amt = int(input("Amount: "))
        balance += amt

    elif ch == 3:
        amt = int(input("Amount: "))
        if amt <= balance:
            balance -= amt
        else:
            print("Insufficient Balance")

    else:
        break
```

```python
students = {}

n = int(input("Enter Number of Students: "))

for i in range(n):
    name = input("Name: ")
    marks = int(input("Marks: "))
    students[name] = marks

average = sum(students.values()) / len(students)

print("\nAverage Marks:", average)

top = max(students, key=students.get)

print("Top Student:", top)
```

```python
import random

secret = random.randint(1, 100)

while True:
    guess = int(input("Guess Number: "))

    if guess == secret:
        print("Correct!")
        break

    elif guess < secret:
        print("Too Low")

    else:
        print("Too High")
```
